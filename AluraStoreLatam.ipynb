### Importación de datos



```python
import pandas as pd
import matplotlib.pyplot as plt


url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda2.head()
```

#1. Análisis de facturación



```python
ingreso_total=0
ingreso_total2=0
ingreso_total3=0
ingreso_total4=0

for loco in tienda["Precio"].astype(float):
  ingreso_total += loco
print(f"Este es el ingreso total de la tienda 1:, {ingreso_total:,}")

for p in tienda2["Precio"].astype(float):
  ingreso_total2 += p
print(f"Este es el ingreso total de la tienda 2:, {ingreso_total2:,}")

for p in tienda3["Precio"].astype(float):
  ingreso_total3 += p
print(f"Este es el ingreso total de la tienda 3:, {ingreso_total3:,}")
for p in tienda4["Precio"].astype(float):
  ingreso_total4 += p
print(f"Este es el ingreso total de la tienda 4:, {ingreso_total4:,}")

def millones(ingreso):
  ingreso=ingreso/1000000000

millones(ingreso_total)
millones(ingreso_total2)
millones(ingreso_total3)
millones(ingreso_total4)

fig, ax = plt.subplots()

tiendas = ['tienda 1', 'tienda 2', 'tienda 3', 'tienda 4']
counts = [1.151, 1.116 ,1.098 , 1.039]
bar_labels = ['tienda 1', 'tienda 2', 'tienda 3', 'tienda 4']
bar_colors = ['tab:red', 'tab:blue', 'tab:green', 'tab:orange']

ax.bar(tiendas, counts, label=bar_labels, color=bar_colors)
plt.yticks([0.0,0.2,0.5,0.8,1 , 1.20])
ax.set_ylabel('Ingreso total (En billones) ')
ax.set_title('Analisis de Facturación')
ax.legend(title='Tiendas')

plt.show()
```

# 2. Ventas por categoría

```python
def ventas_de_categoria(tienda, nombre_tienda):
    ventas_por_categoria = tienda.groupby("Categoría del Producto")["Precio"].sum().reset_index()
    ventas_por_categoria.columns = ['Categoría del Producto', 'Total Vendido']
    ventas_por_categoria = ventas_por_categoria.sort_values(by='Total Vendido', ascending=False)
    relevante = ventas_por_categoria[['Categoría del Producto', 'Total Vendido']]
    print(f"\033[1mVentas por categoria de la {nombre_tienda} \033[0m\n", relevante)
```

```python
ventas_de_categoria (tienda,"Tienda 1")
ventas_de_categoria (tienda2,"Tienda 2")
ventas_de_categoria (tienda3,"Tienda 3")
ventas_de_categoria (tienda4,"Tienda 4")

import matplotlib.pyplot as plt

labels = "Electronicos", '', 'Dogs', 'Logs'
sizes = [15, 30, 45, 10]

fig, ax = plt.subplots()
ax.pie(sizes, labels=labels)
```

```python
def analizar_ventas(tienda):
    dinero_productos = tienda.groupby('Producto')['Precio'].sum().reset_index()
    dinero_productos.columns = ['Producto', 'Total Vendido']
    ventas_productos = tienda['Producto'].value_counts().reset_index()
    ventas_productos.columns = ['Producto', 'Cantidad Vendida']
    resultados = pd.merge(dinero_productos, ventas_productos, on='Producto')
    resultados_relevantes = resultados[['Producto', 'Total Vendido', 'Cantidad Vendida']]
    print("Los 3 productos más vendidos:")
    print(resultados_relevantes.nlargest(3, 'Cantidad Vendida'))
    print("\nLos 3 productos menos vendidos:")
    print(resultados_relevantes.nsmallest(3, 'Cantidad Vendida'))
```
